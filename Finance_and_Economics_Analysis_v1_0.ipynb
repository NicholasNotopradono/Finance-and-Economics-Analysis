```python
import sklearn as sk
import numpy as np
import pandas as pd
import seaborn as sb
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

#importing Libraries

from google.colab import drive
drive.mount('/content/drive')

#mounting in order to connect it to the file

file_path = '/content/drive/MyDrive/Nicks Analyzed Data Sets/finance_economics_dataset.csv' #defining the path to the data set
df = pd.read_csv(file_path)

#finance_economics_dataset.csv will be referred to "df" from now on
```

```python
columns_to_drop = ['Date', 'Stock Index']
df_dropped = df.drop(columns=columns_to_drop)

correlation_matrix = df_dropped.corr()

sb.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")

plt.title("Correlation Matrix Heatmap")
plt.show()

#well this is fucking fun
```
